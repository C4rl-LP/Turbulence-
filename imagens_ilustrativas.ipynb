```python


def bump_function(s, c=1.0):
    """
    Calcula a função bump B(s) = exp(-c / (1 - s^2)) para |s| < 1
    e retorna 0 para |s| >= 1.
    """
    # Cria um array de zeros com o mesmo tamanho de s
    B = np.zeros_like(s)
    
    # Encontra os índices onde a condição |s| < 1 é verdadeira
    valid_indices = np.abs(s) < 1.0
    
    # Aplica a fórmula apenas nos índices válidos para evitar divisão por zero
    B[valid_indices] = np.exp(-c / (1.0 - s[valid_indices]**2))
    
    return B

# Define o domínio de s (de -1.5 a 1.5 para mostrar o decaimento a zero nas bordas)
s_values = np.linspace(-1.5, 1.5, 1000)

# Valores de c para comparar
c_values = [0.7, 1.0, 2.0, 5.0]

# Configuração do gráfico
plt.figure(figsize=(10, 6))

for c in c_values:
    B_values = bump_function(s_values, c)
    plt.plot(s_values, B_values, label=f'c = {c}', linewidth=2)

# Estilização do gráfico
plt.title('Função Bump $B(s)$ para diferentes valores de c', fontsize=14)
plt.xlabel('x', fontsize=12)
plt.ylabel('B(x)', fontsize=12)
plt.axhline(0, color='black', linewidth=1) # Eixo x
plt.axvline(0, color='black', linewidth=1) # Eixo y
plt.axvline(-1, color='red', linestyle='--', alpha=0.5, label='Fronteiras |x| = 1')
plt.axvline(1, color='red', linestyle='--', alpha=0.5)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=11)

# Exibe o gráfico
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

fig, axs = plt.subplots(1, 3, figsize=(12, 4))

# calcula limites globais
todos = []

for n in [1, 2, 3]:
    centros = centros_nivel_todos(n, 0)
    todos.append(centros)

todos = np.vstack(todos)

xmin, xmax = todos[:,0].min(), todos[:,0].max()
ymin, ymax = todos[:,1].min(), todos[:,1].max()

# mesma borda
margem = 0.1 * max(xmax - xmin, ymax - ymin)

xmin -= margem
xmax += margem
ymin -= margem
ymax += margem

for i, n in enumerate([1, 2, 3]):

    centros = centros_nivel_todos(n, 0)

    axs[i].scatter(centros[:,0], centros[:,1])

    axs[i].set_title(f"n = {n}")

    # limites idênticos
    axs[i].set_xlim(xmin, xmax)
    axs[i].set_ylim(ymin, ymax)

    # mesma escala física
    axs[i].set_aspect('equal', adjustable='box')

    axs[i].grid(True)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(6, 6))

# guardar todos os pontos para calcular limites globais
todos = []

for n in [1, 2, 3]:

    centros = centros_nivel_todos(n, 0.02)

    todos.append(centros)

    ax.scatter(
        centros[:,0],
        centros[:,1],
        label=f"nível {n}"
    )

# limites iguais em x e y
todos = np.vstack(todos)

xmin, xmax = todos[:,0].min(), todos[:,0].max()
ymin, ymax = todos[:,1].min(), todos[:,1].max()

margem = 0.1 * max(xmax - xmin, ymax - ymin)

ax.set_xlim(xmin - margem, xmax + margem)
ax.set_ylim(ymin - margem, ymax + margem)

ax.set_aspect('equal', adjustable='box')

ax.set_xlabel("x")
ax.set_ylabel("y")

ax.legend()
ax.grid(True)

plt.tight_layout()
plt.show()
```
